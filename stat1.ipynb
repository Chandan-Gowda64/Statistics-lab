```python
import pandas as pd
import matplotlib.pyplot as plt
data={
    "Name":["A","B","c","D","E"],
    "Marks":[65,72,81,55,98]
}
df=pd.DataFrame(data)
print(df)
print(df["Marks"].mean())
plt.bar(df["Name"],df["Marks"])
plt.xlabel("Student")
plt.ylabel("Marks")
plt.title("Student marks")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

result=np.random.choice(["Head","Tail"],size=1000)
heads=np.sum(result=="Head")

print("Heads:",heads)
print("Probabilty of Head:",heads/1000)
counts=[
    np.sum(result=="Head"),
    np.sum(result=="Tail")
]
plt.bar(["Head","Tail"],counts)
plt.xlabel("Outcome")
plt.ylabel("Count")
plt.title("coin toss")
plt.show()
```
